2.Betimsel İstatistikler (Tek Değişkenli Analiz)


1.soru :Sayısal değişkenler (yas, siparis_tutari_tl, siparis_adedi, memnuniyet_puani) için ortalama, medyan, standart sapma, minimum ve maksimum değerleri hesaplayın.

```python
sayisal_degiskenler = ['yas', 'siparis_tutari_tl', 'siparis_adedi', 'memnuniyet_puani']
df[sayisal_degiskenler].describe()

#sayısal degıskenlerı bulmak
```

2,soru:Sayısal değişkenler (yas, siparis_tutari_tl, siparis_adedi, memnuniyet_puani) için ortalama, medyan, standart sapma, minimum ve maksimum değerleri hesaplayın.

```python
from numpy import select
selected_columns = df.select_dtypes(include=['object']).columns.tolist()
selected_columns = selected_columns[1:]

for column in selected_columns:
  unique_values = df[column].unique()
  print(f"Unique values in {column}: {unique_values}\n")
  print("-"*50)
```

3.Soru : Kategorik değişkenlerin frekans (adet) ve yüzde dağılımlarını listeleyin

```python
for col in selected_columns:
    print("\nSütun:", col)

    print("Frekans (Adet):")
    frekans = df[col].value_counts()
    print(frekans)

    print("\nYüzde Dağılımı (%):")
    yuzde = df[col].value_counts(normalize=True) * 100
    print(yuzde)
```

3. Değişkenler Arası İlişkiler ve Gruplandırma

1.Soru: Şehirlere göre gruplandırma yaparak her şehrin ortalama sipariş tutarını ve ortalama memnuniyet puanını hesaplayın.

```python
df.groupby("sehir")[["siparis_tutari_tl","memnuniyet_puani"]].mean()
# sehırlerın ort sıparısntutarı ve ort memnunıyet puanı
```

2.Soru: Ürün kategorilerine göre toplam sipariş adetlerini karşılaştırın.

```python
df.groupby("urun_kategorisi")["siparis_adedi"].sum()
# Ürün kategorilerine göre toplam sipariş adetleri
```

3.Soru: Cinsiyet ve eğitim düzeyine göre iade oranlarını incelemek için bir çapraz tablo (crosstab) veya pivot tablo oluşturun.

```python
tablo = pd.crosstab(df["cinsiyet"], df["egitim_duzeyi"], values=(df["iade_durumu"] == "Evet"), aggfunc="mean")
print(tablo) # Cinsiyet ve eğitim düzeyine göre cross tablo ıade durumu
```

4.Soru: Müşteri yaşı ile sipariş tutarı arasında istatistiksel olarak bir korelasyon (ilişki) olup olmadığını hesaplayın.

```python
korelasyon = df['yas'].corr(df['siparis_tutari_tl'])
print(f"Müşteri yaşı ile sipariş tutarı arasındaki korelasyon: {korelasyon}")

#Müşteri yaşı ile sipariş tutarı arasında korelasyon ilişkisi
```

4. Veri Görselleştirme

```python
import matplotlib.pyplot as plt
```

```python
1.Soru: En çok sipariş verilen ürün kategorilerini Sütun Grafiği (Bar Chart) ile gösterin.
```

```python
# Ürün kategorilerinin sipariş sayısını hesaplatmak
kategori_sayilari = df["urun_kategorisi"].value_counts()

# Sütun grafiğinı oluşturmak
kategori_sayilari.plot(kind="bar")

plt.title("En Çok Sipariş Verilen Ürün Kategorileri")
plt.xlabel("Ürün Kategorisi")
plt.ylabel("Sipariş Sayısı")

plt.show()
```

2.Soru: Şehirlerin müşteri dağılımındaki payını Pasta Grafiği (Pie Chart) ile çizin.

```python
# Şehirlere göre müşteri sayısını hesaplamak
sehir_dagilimi = df["sehir"].value_counts()

# Pasta grafiği oluşturmak
plt.pie(sehir_dagilimi, labels=sehir_dagilimi.index)
#Pasta Grafiği (Pie Chart) çizmek için

plt.title("Şehirlere Göre Müşteri Dağılımı")

plt.show()
```

3.Soru: Müşterilerin yaş dağılımını ve sipariş tutarı dağılımını Histogram kullanarak görselleştirinMüşterilerin yaş dağılımını ve sipariş tutarı dağılımını Histogram kullanarak görselleştirin

```python
# Müşteri yaşı histogramı
plt.hist(df["yas"], bins=10)  #hist =Histogram grafiği çizmek için

plt.title("Müşteri Yaşı Dağılımı")
plt.xlabel("Yaş")
plt.ylabel("Müşteri Sayısı")

plt.show()


# Sipariş tutarı histogramı
plt.hist(df["siparis_tutari_tl"], bins=10)

plt.title("Sipariş Tutarı Dağılımı")
plt.xlabel("Sipariş Tutarı (TL)")
plt.ylabel("Sipariş Sayısı")

plt.show()
```

4.Soru: Sipariş tutarındaki aykırı değerleri (outlier) gözlemlemek için bir Kutu Grafiği (Boxplot) çizin.

```python
# Sipariş tutarı için boxplot çizme
plt.boxplot(df["siparis_tutari_tl"])

plt.title("Sipariş Tutarı Aykırı Değer Analizi")
plt.ylabel("Sipariş Tutarı (TL)")

plt.show()
```

5.Soru: Yaş ve sipariş tutarı değişkenlerini bir Serpilme Diyagramı (Scatter Plot) üzerinde göstererek aralarındaki ilişkiyi görselleştirin

```python
# Scatter plot çizme
plt.scatter(df["yas"], df["siparis_tutari_tl"])
#scatter = Serpilme Diyagramı (Scatter Plot) çizmek için

plt.title("Yaş ve Sipariş Tutarı Arasındaki İlişki")
plt.xlabel("Müşteri Yaşı")
plt.ylabel("Sipariş Tutarı (TL)")

plt.show()
```
